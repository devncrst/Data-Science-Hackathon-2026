# Score new data with our SAVED models (no retraining)
This notebook only **loads** the trained models in `models/`. Nothing is retrained.

Steps inside `predict.predict_city`: the same cleaning and features as in training, neighbourhood features (our own pixels of that city are used as extra neighbours), the saved model, 5x5 smoothing. Every input pixel gets exactly one prediction. If the file contains `weighted_mean_year`, the macro F1 is printed for checking.

**Set the three values below, then Run All.** The default `TEST_FILE` is our own Amsterdam training file, only to show that it works: that score is NOT a fair test (those pixels were used in training).

```python
import predict

TEST_FILE   = "data/amsterdam_data.parquet"   # <- path to the held-out raw parquet file
CITY        = "Amsterdam"                     # "Amsterdam" or "Madrid"
N_PER_CLASS = 25                              # Amsterdam model to use: 5, 25, 50, 100 or 200 labels per class

amsterdam_model = f"models/amsterdam_blend3_n{N_PER_CLASS:03d}_seed42.joblib" if CITY == "Amsterdam" else None
out = predict.predict_city(TEST_FILE, city=CITY,
                           stage1_path="models/stage1_madrid_model.joblib",   # saved Stage 1 (Madrid) model
                           amsterdam_model_path=amsterdam_model)              # saved Stage 2 model (Amsterdam only)
name = f"predictions_{CITY.lower()}" + (f"_n{N_PER_CLASS:03d}" if CITY == "Amsterdam" else "") + ".csv"
out.to_csv(name, index=False)
print("saved", name, "with", len(out), "pixels")
out.head()
```
